```python
from azure.identity import DefaultAzureCredential
from azure.storage.blob import BlobServiceClient

import json
```

```python
try:
    account_url = "https://playtactixstorage.blob.core.windows.net"
    container_name = "skillcorner-data"
    blob_name = "hello-world.json"
    default_credential = DefaultAzureCredential()

    # Create the BlobServiceClient object
    blob_service_client = BlobServiceClient(account_url, credential=default_credential)
    blob_client = blob_service_client.get_blob_client(container=container_name, blob=blob_name)
    
    payload = {
        "message": "Hello, World!"
    }
    
    blob_client.upload_blob(json.dumps(payload), overwrite=True)
    print(f"Blob '{blob_name}' uploaded successfully to container '{container_name}'.")
    
except Exception as ex:
    print('Exception:')
    print(ex)
```

```python
import pandas as pd

from io import BytesIO
parquet_file = '../../data/silver_tracking_data.parquet'
# df = pd.read_parquet(parquet_file)
```

```python
try:
    account_url = "https://playtactixstorage.blob.core.windows.net"
    container_name = "skillcorner-data"
    blob_name = "uploaded-test.parquet"
    default_credential = DefaultAzureCredential()

    # Create the BlobServiceClient object
    blob_service_client = BlobServiceClient(account_url, credential=default_credential)
    blob_client = blob_service_client.get_blob_client(container=container_name, blob=blob_name)
    
    df = pd.read_parquet(parquet_file)
    parquet_buffer = BytesIO()

    df.to_parquet(parquet_buffer, index=False)

    parquet_buffer.seek(0)
    
    blob_client.upload_blob(parquet_buffer, overwrite=True)
    print(f"Blob '{blob_name}' uploaded successfully to container '{container_name}'.")
    
except Exception as ex:
    print('Exception:')
    print(ex)
```

```python
try:
    account_url = "https://playtactixstorage.blob.core.windows.net"
    container_name = "skillcorner-data"
    blob_name = "uploaded-test.parquet"
    default_credential = DefaultAzureCredential()

    # Create the BlobServiceClient object
    blob_service_client = BlobServiceClient(account_url, credential=default_credential)
    blob_client = blob_service_client.get_blob_client(container=container_name, blob=blob_name)
    
    downloaded_blob = blob_client.download_blob()
    # Read bytes

    parquet_bytes = downloaded_blob.readall()

    # Convert bytes to DataFrame

    parquet_buffer = BytesIO(parquet_bytes)

    df = pd.read_parquet(parquet_buffer)

    print(

        f"Blob '{blob_name}' downloaded successfully "

        f"from container '{container_name}'."

    )

    print(type(df))

    print(df.head())
    
except Exception as ex:
    print('Exception:')
    print(ex)
```
